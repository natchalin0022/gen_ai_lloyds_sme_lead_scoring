```python
URL_BASE = "https://api.company-information.service.gov.uk"
URL_META_DATA = "https://document-api.company-information.service.gov.uk/document"
TIMEOUT = 10  # seconds
```

```python
def get(endpoint, params=None, headers=None):
    url = f"{URL_BASE}/{endpoint}"
    response = requests.get(
        url,
        auth=(os.getenv("CH_API"),""),
        headers=headers,
        params=params,
        timeout=TIMEOUT
        )
    response.raise_for_status()
    print("start_index:", params.get("start_index") if params else None)
    print(f"{endpoint} - {response.status_code}")
    return response.json()
```

```python
# combination of start_index and size under under 10,000 index limit

response = get("advanced-search/companies", params={"start_index": 9980, "size":20})
response["items"][0]
```
