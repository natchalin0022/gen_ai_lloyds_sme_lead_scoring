```python
from matplotlib.pyplot import subplots
import csv
import pandas as pd
from pathlib import Path
```

```python
with open(Path.cwd() / 'Qwen3_VL_8B_fine_tuning.csv', 'r') as file:
    data = pd.read_csv(file)
    data.set_index("step", inplace=True)

data
```

```python
ax = subplots(figsize=(5, 5))[1]
ax.plot(data.index, data['training_loss'], label='Training Loss', color='blue')
ax.plot(data.index, data['validation_loss'], label='Evaluation Loss', color='orange')
ax.legend(fontsize=14)
```
