```python
import os
from dotenv import load_dotenv, dotenv_values
from pathlib import Path
import requests
import pandas as pd

load_dotenv(Path("/Users/tanaphan/Documents/STUDY/UoBristol/Final_Project/team_21_lloyds/.env"))
```

```python
API_KEY = os.getenv("API_KEY_GUARDIAN")
```

```python
from bs4 import BeautifulSoup

def html_remover(raw):
    if not raw:
        return ""
    soup = BeautifulSoup(raw, "html.parser")
    html_cleaned = soup.get_text()
    return html_cleaned
```

```python
def pull_news(KEYWORD):
    response = requests.get(
    "https://content.guardianapis.com/search",
    params={
        "q": KEYWORD,
        "api-key": API_KEY,
        "page-size": 50,
        "order-by": "newest",
        # "section": "business",
        "show-fields": "trailText,headline,body,wordcount,lastModified",
        "show-tags": "keyword"
    })
    
    data = response.json()["response"]
    articles = data.get("results", [])

    print(f"Status: {data['status']}")
    print(f"Total Articles: {data['total']}")
    print(f"Total Pages: {data['pages']}")

    rows = []
    for article in articles:
        fields = article.get("fields", {})
        tags = article.get("tags", [])
        tag_names = []
        for tag in tags:
            tag_names.append(tag["webTitle"])

        rows.append({
            "title": article.get("webTitle", ""),
            "section": article.get("sectionName", ""),
            "date": article.get("webPublicationDate", ""),
            "url": article.get("webUrl", ""),
            "trail_text": fields.get("trailText", ""),
            "wordcount": fields.get("wordcount", None),
            "tags": tag_names,
            "body": html_remover(fields.get("body", ""))
        })

    df = pd.DataFrame(rows)
    return df
```

```python
df = pull_news('"Lloyds Bank" AND SME')
```
