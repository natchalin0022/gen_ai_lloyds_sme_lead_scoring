# Filter for samples

25 -> positives
75 -> negatives

Scored

```python
import pandas as pd

df = pd.read_csv('/Users/natchalin_/Projects/final_project/Lloyds/API/flat_pot.csv')
```

```python
df_pos = df[df['label'] == 1].sample(n = 25, random_state = 40)
```

```python
df_neg = df[df['label'] == 0].sample(n = 75, random_state = 40)
```

```python
df_sample = pd.concat([df_pos, df_neg], ignore_index = True)
```

```python
df_sample['label'].value_counts()
```

```python
df_sample = df_sample[['com_num', 'name', 'label']]
```

```python
df_sample['relationship'] = 'prospect'
```

```python
df_sample_final = df_sample.drop(columns = ['label'])
```

```python
df_sample_final = df_sample_final.rename(columns = {'com_num': 'company_number', 'name': 'company_name'})
```

```python
df_sample_final.to_csv('sample.csv')
```

# Combine the scores

```python
scored = pd.read_csv('leads copy.csv')
```

```python
scored = scored[['com_num', 'name', 'gated']]
```

```python
merged = scored.merge(df_sample_final, left_on='com_num', right_on='company_number', how='left')
```

```python
merged = merged[['com_num', 'name', 'gated','relationship']]
```
