# Random 50 company numbers

Picks 50 random companies from `30k_sme_for_mock_up.csv` and saves them as a new CSV (one column, `company_number`) in this folder. Each run writes a new timestamped file.

```python
from datetime import datetime
from pathlib import Path

import pandas as pd

FOLDER = Path.cwd()
SOURCE = FOLDER / "30k_sme_for_mock_up.csv"
N = 50

# read as text so IDs like SC123456 / 00012345 keep their letters and leading zeros
ids = pd.read_csv(SOURCE, dtype=str, usecols=["company_id"])["company_id"].dropna().drop_duplicates()

sample = ids.sample(n=N).rename("company_number")

out_path = FOLDER / f"random_{N}_companies_{datetime.now():%Y%m%d_%H%M%S}.csv"
sample.to_csv(out_path, index=False)

print(f"Saved {len(sample)} company numbers -> {out_path.name}")
sample.head()
```
